```python
from sklearn.tree import DecisionTreeClassifier
# step1 : collect data .
# Feature : [Weight (gram),colour(0=red,1=orange)]
```

```python
feature=[[150,0],#apple
         [160,0],#apple
         [170,1],#orange
         [180,1]]#orange
```

```python
labels=["apple","apple","orange","orange"]
```

```python
#step2 :Train the modal .
```

```python
clf = DecisionTreeClassifier()
clf=clf.fit(feature,labels)
```

```python
prediction =clf.predict([[150,0],[180,1]])
```

```python
print(f"those are {prediction[0],prediction[1]}")
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
plt.figure(figsize=(10,6))
plot_tree(clf, feature_names=["weight", "color"], class_names=["apple", "orange"], filled=True)
plt.tight_layout()
plt.show()
```
