```python
import pandas as pd 
from sklearn.ensemble import RandomForestClassifier
```

```python
data=pd.read_csv("iris-train.csv")
# data.to_csv("iris-train.csv",index=False) #to convert file in csv
```

```python
data.head()
```

```python
X = data.iloc[:, :-1]  # All columns except the last as features
y = data.iloc[:, -1]   # Last column as label
```

```python
# Train the Model
model=RandomForestClassifier()
model.fit(X,y)
print("Number of features:", X.shape[1])
# The model is now trained on your complete dataset.
```

```python
# Inference

predictions = model.predict([[7.9,	3.8,	6.4	,2]]) 
print(predictions)  # Display predictions




```

```python
# data = data.drop(columns=['Unnamed: 0.2'])#this is use to remove the unwanted row or column.
```

```python
predictions = model.predict(X)  # Predict on the same/new data (for demonstration)
print(predictions)  # Display predictions
```

```python
x_test=datatest.iloc[:,:-1]
```

```python
predictions1=model.predict(x_test.values)
predictions1
```

```python
y_test=datatest.iloc[:,-1]
```

```python
actual=y_test.values
```

```python
# lets find accuracy.
# In simple terms, accuracy tells us how often the model was right.
# If you gave your model 100 questions to answer, and it got 90 of them correct, its accuracy would be:
# Accuracy = Correct Predictions / Total Predictions = 90 / 100 = 90%
```

```python
# now,
count=0
for i in range (0,len(actual)):
    if actual[i]==predictions1[i]:
        count=count+1
print((count*100)/len(actual))
```
