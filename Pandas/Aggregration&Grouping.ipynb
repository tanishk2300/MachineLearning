```python
import pandas as pd 
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
```

```python
# here we can see the given data , now we are using the grouping .
```

```python
df
```

```python
# here we get the avg salary as per department .
```

```python
 df.groupby("Team")["Salary"].mean()  # Average per team
```

```python
df.groupby("Team")["Salary"].sum()      # Total score
```

```python
df.groupby("Team")["Salary"].count()    # How many entries
```

```python
df.groupby("Team")["Salary"].min()# here the min salary was show 
```

```python
df.groupby("Team")["Salary"].max() # here the max salary show 
```

```python
df.groupby("Team")["Salary"].agg(["min","max"]) # this help to applying the multiple function at once .
```

```python
df.groupby("Team")["Salary"] .agg(avg_score="mean",high_Score="max")
```

```python
df.groupby("Team").agg({"Salary":"mean","Age":"max"})
```

```python
df["Team Avg"]=df.groupby("Team")["Salary"].transform("mean").copy()
```
